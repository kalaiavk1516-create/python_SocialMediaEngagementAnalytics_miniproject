```
df= df.drop_duplicates()
```

```
df["gender"] =df["gender"].str.strip().str.title()
```

```
numeric_cols=[
    "likes",
    "comments",
    "shares",
    "watch_time_sec",
    "impression_count",
    "follower_count"
]

df[numeric_cols] = df[numeric_cols].apply(
    pd.to_numeric,
    errors="coerce")
```

```
df[["likes","comments","shares"]].describe()
```

```
df.loc[df["likes"]<0,"likes"] = 0
df.loc[df["comments"]<0,"comments"] = 0
df.loc[df["shares"]<0,"shares"] = 0
```

```
# foodie # travel #love

df["hastag_count"]= df["hashtags"].fillna("").apply(lambda x: len(x.split(",")))
```

```
df["sentiment"] = df["sentiment"].str.strip().str.lower()
```

```
import matplotlib.pyplot as plt
import seaborn as sns
```

```
plt.scatter(df["impression_count"], df["likes"])
plt.xlabel("Impression Count")
plt.ylabel("Likes")
plt.title("Likes vs Impression Count")
plt.show()
```

```
df["post_category"].value_counts().plot(kind="bar")
plt.xlabel("Category")
plt.ylabel("number of post")
plt.title("Posts by Category")
plt.show()
```

```
df["gender"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Gender Distribution")
plt.ylabel("")
plt.show()
```

```
sns.barplot(data=df,
    x="post_category",
    y="likes"
)
plt.title("average likes by category")
plt.xticks(rotation=45)
plt.show()
```
